```python
import json

import pandas as pd
import numpy as np

from scipy.sparse import lil_matrix

import matplotlib.pyplot as plt

from collections import Counter
```

```python

movies = []
with open("imdb_movies_2000to2022.prolific.json", "r") as in_file:
    for line in in_file:
        
        # Read the movie on this line and parse its json
        this_movie = json.loads(line)
        movies.append(this_movie)


print(f"Read {len(movies)} movies from the file")
```

```python
def jaccard_similarity(actors1, actors2):
    """
    Calculate the Jaccard similarity between two sets of actors.
    """
    intersection = actors1.intersection(actors2)
    union = actors1.union(actors2)
    return len(intersection) / len(union) if union else 0

def predict_genre(target_movie_id, k):
    for movie in movies:
        if movie["imdb_id"] == target_movie_id:
            print(movie["imdb_id"])
            target_movie = movie
            break
    target_actors = {actor[0] for actor in target_movie["actors"]}
    similarities = []
    for movie in movies:
        if movie["imdb_id"] != target_movie_id:
            actors = {actor[0] for actor in movie["actors"]}
            similarity = jaccard_similarity(target_actors, actors)
            similarities.append((similarity, movie["genres"]))
    similarities.sort(reverse=True, key=lambda x: x[0])
    top_k_genres = []
    for similarity, genres in similarities[:k]:
        for genre in genres:
            top_k_genres.append(genre)
    genre_counts = Counter(top_k_genres)
    max_count = max(genre_counts.values())
    most_common_genres = [genre for genre, count in genre_counts.items() if count == max_count]
    return most_common_genres
    
```

```python
target_movie_id_1 = "tt0317705" # The Incredibles
target_movie_id_2 = "tt0816692" # Interstellar
target_movie_id_3 = "tt0332280" # The Notebook

k_1 = 1
k_3 = 3
k_5 = 5
predicted_genres = predict_genre(target_movie_id_1, k_1)
print(f"Predicted genres for movie 1 with k=1 {target_movie_id_1}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_1, k_3)
print(f"Predicted genres for movie 1 with k=3 {target_movie_id_1}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_1, k_5)
print(f"Predicted genres for movie 1 with k=5 {target_movie_id_1}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_2, k_1)
print(f"Predicted genres for movie 2 with k=1 {target_movie_id_2}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_2, k_3)
print(f"Predicted genres for movie 2 with k=3 {target_movie_id_2}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_2, k_5)
print(f"Predicted genres for movie 2 with k=5 {target_movie_id_2}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_3, k_1)
print(f"Predicted genres for movie 3 with k=1 {target_movie_id_3}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_3, k_3)
print(f"Predicted genres for movie 3 with k=3 {target_movie_id_3}: {predicted_genres}")
predicted_genres = predict_genre(target_movie_id_3, k_5)
print(f"Predicted genres for movie 3 with k=5 {target_movie_id_3}: {predicted_genres}")
```
