```python
import pandas as pd
import numpy as np
```

```python
model1_df = pd.read_csv('prediction_model_01.csv')
model2_df = pd.read_csv('prediction_model_02.csv')
model3_df = pd.read_csv('prediction_model_03.csv')

def calculate_metrics(df, genre):
    true_positive = 0
    false_positive = 0
    false_negative = 0
    true_negative = 0

    for _, row in df.iterrows():
        predicted_genre = row['predicted']
        actual_genres = eval(row['actual genres'])  
        correct = row['correct?']

        if genre in actual_genres:
            if predicted_genre == genre:
                true_positive += 1
            else:
                false_negative += 1
        else:
            if predicted_genre == genre:
                false_positive += 1
            else:
                true_negative += 1

    accuracy = (true_positive + true_negative) / len(df)
    precision = true_positive / (true_positive + false_positive) if (true_positive + false_positive) > 0 else 0
    recall = true_positive / (true_positive + false_negative) if (true_positive + false_negative) > 0 else 0
    f1_score = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    return accuracy, precision, recall, f1_score

genres_to_evaluate = ['Drama', 'Comedy', 'Horror']
models = {'Model 1': model1_df, 'Model 2': model2_df, 'Model 3': model3_df}

results = {}
for model_name, df in models.items():
    results[model_name] = {}
    for genre in genres_to_evaluate:
        metrics = calculate_metrics(df, genre)
        results[model_name][genre] = {
            'Accuracy': metrics[0],
            'Precision': metrics[1],
            'Recall': metrics[2],
            'F1 Score': metrics[3]
        }

#  results
for model_name, metrics in results.items():
    print(f"Metrics for {model_name}:")
    for genre, genre_metrics in metrics.items():
        print(f"  {genre}: {genre_metrics}")
    print()
```

```python
model1_df = pd.read_csv('prediction_model_01.csv')

genres_to_evaluate = ['Drama', 'Comedy', 'Horror']

genre_true_counts = {genre: 0 for genre in genres_to_evaluate}
genre_tp_counts = {genre: 0 for genre in genres_to_evaluate}
genre_fp_counts = {genre: 0 for genre in genres_to_evaluate}

for idx, row in model1_df.iterrows():
    actual_genres = eval(row['actual genres'])

    for true_g in genres_to_evaluate:
        if true_g in actual_genres:
            genre_true_counts[true_g] += 1
    predicted_genre = row['predicted']
    correct = row['correct?']

    for genre in genres_to_evaluate:
        if predicted_genre == genre and correct == 1:
            genre_tp_counts[genre] += 1
        elif predicted_genre == genre and correct == 0:
            genre_fp_counts[genre] += 1


precision = {}
recall = {}
f1_score = {}
accuracy = 0
correct_predictions = 0

for genre in genres_to_evaluate:
    tp = genre_tp_counts[genre]
    fp = genre_fp_counts[genre]
    fn = genre_true_counts[genre] - tp

    if tp + fp > 0:
        precision[genre] = tp / (tp + fp)
    else:
        precision[genre] = 0

    if tp + fn > 0:
        recall[genre] = tp / (tp + fn)
    else:
        recall[genre] = 0

    if precision[genre] + recall[genre] > 0:
        f1_score[genre] = (2 * precision[genre] * recall[genre]) / (precision[genre] + recall[genre])
    else:
        f1_score[genre] = 0

    accuracy += tp + (len(model1_df) - genre_true_counts[genre] - fp)
    correct_predictions += tp

accuracy = correct_predictions / len(model1_df)
print(f"Accuracy: {accuracy:.4f}\n")
print("Precision:")
for genre, prec in precision.items():
    print(f"  {genre}: {prec:.4f}")
print("\nRecall:")
for genre, rec in recall.items():
    print(f"  {genre}: {rec:.4f}")
print("\nF1 Score:")
for genre, f1 in f1_score.items():
    print(f"  {genre}: {f1:.4f}")

    

```
