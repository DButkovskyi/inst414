```python
import pandas as pd
import json
```

```python
df = pd.read_csv('movie_to_cluster.csv')
df.head()
```

```python
movie_genre_map = {}

with open('imdb_movies_2000to2022.prolific.json') as f:
    for line in f:
        movie = json.loads(line)
        movie_genre_map[movie['imdb_id']] = movie['genres']
```

```python
# Combine df and map based on id into a new dataframe where df movie_id = map imdb_id, make genres as comma separated string

df['genres'] = df['movie_id'].map(movie_genre_map)
df['genres'] = df['genres'].apply(lambda x: ', '.join(x))

df.head()
```

```python
# Set your target genre as "Sci-Fi"
target_genre = "Sci-Fi"

total_movies = len(df)
print(total_movies)

sci_fi_movies = len(df[df['genres'].str.contains(target_genre)])
print(sci_fi_movies)

clusters = df['cluster'].unique()
print(clusters)


```

```python
# Use a for loop to iterate through the clusters. For each one, calculate the probability that a new movie will 
# fit in that cluster given that it is a Sci-Fi movie. In other words calculate P(Cluster | Sci-Fi)

for cluster in clusters:
    cluster_movies = len(df[df['cluster'] == cluster])
    sci_fi_in_cluster = len(df[(df['cluster'] == cluster) & (df['genres'].str.contains(target_genre))])
    prob = sci_fi_in_cluster / sci_fi_movies
    print(f"Probability of {cluster} given Sci-Fi: {prob}")
```

```python
#3. Set your target cluster as "Cluster0"

target_cluster = 0

#4. Use a for loop to iterate through the genres. For each one,  calculate P(Genre | Cluster0) 
# Genres cintains list of genres comma separated and I only look for individual genres

for genre in df['genres'].str.split(', ').explode().unique():
    if genre.strip() == "":
        continue
    genre_movies = len(df[(df['genres'].str.contains(genre)) & (df['cluster'] == target_cluster)])
    cluster_movies = len(df[df['cluster'] == target_cluster])
    if cluster_movies == 0:
        prob = 0
    else:
        prob = genre_movies / cluster_movies
    print(f"Probability of {genre} given {target_cluster}: {prob}")
```
